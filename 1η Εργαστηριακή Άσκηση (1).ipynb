```python
import numpy as np
import matplotlib.pyplot as plt

#Οριζούμε Συνάρτηση f(x)
def f(x):
    return 1 / (1 + 25 * x**2)

#Tο πολύωνυμο lagrange απο τις σημειώσεις
def lagrange_interpolation(x_nodes, f, x):
    n = len(x_nodes)
    result = 0.0
    for i in range(n):
        # Υπολογισμός του L_i(x)
        L_i = 1.0
        for j in range(n):
            if i != j:
                L_i *= (x - x_nodes[j]) / (x_nodes[i] - x_nodes[j])
        result += f[i] * L_i
    return result


#Οι κόμβοι παρεμβολής x_data να είναι ισαπέχοντα σημεία
#, θα χρησιμοποιήσουμε linspace
# Διάστημα x ∈ [-1, 1]
x_plot = np.linspace(-1, 1, 100) #χωρισμένο σε 100 υποδιαστήματα
#f(x):
fx=f(x_plot)

#Για το πολυώνυμο 6ου βάθμου
#ισαπέχοντα σημεία
p6_x=np.linspace(-1,1,7)#χρειαζόμαστε 7 σημεία για το πολυώνυμο 6 βαθμού
p6_y=f(p6_x)
print("Τα σημεία είναι p6 : ",p6_x)
print("Τα σημεία f(x) p6 είναι: ",p6_y)
#υπολογισμός πολυώνυμου του 6ου βαθμου
p6_lagrange = np.zeros(len(x_plot))
for i in range(len(x_plot)):
    p6_lagrange[i] = lagrange_interpolation(p6_x, p6_y, x_plot[i])


print(p6_lagrange)

#Για το πολυώνυμο 20 ου βάθμου
#ισαπέχοντα σημεία
p20_x=np.linspace(-1,1,21)#χρειαζόμαστε 21 σημεία για το πολυώνυμο 20 βαθμού
p20_y=f(p20_x)
print("Τα σημεία είναι p20: ",p20_x)
print("Τα σημεία f(x) p20 είναι: ",p20_y)
#υπολογισμός πολυώνυμου του 20ου βαθμου
p20_lagrange = np.zeros(len(x_plot))
for i in range(len(x_plot)):
    p20_lagrange[i] = lagrange_interpolation(p20_x, p20_y, x_plot[i])

print(p20_lagrange)

#Γραφική της (f,p6,p20)
plt.figure(figsize=(10,6))
plt.gca().set_facecolor('lightyellow')
#η f(x) μας
plt.plot(x_plot, fx, color='blue', label="f(x)")
#το πολυώνυμο lagrange 6oυ βαθμού με  plot: x_plot, σημεία παρεμβολής p6
plt.plot(x_plot, p6_lagrange, color='red', label="p6")
#το πολυώνυμο lagrange 20oυ βαθμού με  plot: x_plot, σημεία παρεμβολής p20
plt.plot(x_plot, p20_lagrange, color='green', label="p20")
plt.title("Γραφική παράσταση της (f,p6,p20)")
plt.grid(True)
plt.legend()
plt.xlabel(" άξονας x")
plt.ylabel("άξονας y")
plt.show()



```

```python
#Για την χρήση των σημείων Chebyshev έχουμε ότι
#𝑘 = 0,1, … ,20
k = np.arange(21)
print(k)

#για τον τύπο με το cos και τα x
Chebyshev_x=  np.cos((2*k + 1)*np.pi/42)
print("Chebyshev x : ", Chebyshev_x)
#επομένως οι τιμές της f ειναι της μορφής
Chebyshev_y= f(Chebyshev_x)
print("Chebyshev f(x) : ",Chebyshev_y)

#σαφώς απο πρίν εχούμε το lagrange_interpolatio
# επομένως ο υπολογισμός του q20 είναι

q20_lagrange = np.zeros(len(x_plot))
for i in range(len(x_plot)):
    q20_lagrange[i] = lagrange_interpolation(Chebyshev_x, Chebyshev_y, x_plot[i])

print(q20_lagrange)

#Γραφική της (f,q20)
plt.figure(figsize=(10,6))
plt.gca().set_facecolor('lightpink')
#η f(x) μας
plt.plot(x_plot, fx, color='magenta', label="f(x)")
#το πολυώνυμο lagrange (Chebyshev) 20oυ βαθμού
# με  plot: x_plot, σημεία παρεμβολής q20
plt.plot(x_plot, q20_lagrange, color='black', label="q20")
plt.title("Γραφική παράσταση της (f,q20)")
plt.grid(True)
plt.legend()
plt.xlabel(" άξονας x")
plt.ylabel("άξονας y")
plt.show()


```
